```python
class Employee:
    def __init__(self,name,salary):
        self.name=name
        self.salary=salary
    def display_details(self):
        print('employee name:',self.name)
        print('employee salary:',self.salary)
class Manager(Employee):
    def __init__(self, name, salary,team_size):
        super().__init__(name, salary)
        self.team_size=team_size
    def display_role(self):
        print('manager team size is', self.team_size)
class Engineer(Employee):
    def __init__(self, name, salary,specilization):
        super().__init__(name, salary)
        self.specilization=specilization
    def display_role(self):
        print('engineer specialization is', self.specilization)

mgr=Manager('alice',80000,10)
mgr.display_details()   
mgr.display_role()
print('-'*30)
eng=Engineer('bob',60000,'Python')
eng.display_details()
eng.display_role()
```

```python
class payment:
    def __init__(self,amount):
        self.amount=amount
    def process_payment(self):
        print('processing payment of amount:',self.amount)
class credit_card_payment(payment):
    def process_payment(self):
        print('processing credit card payment of amount:',self.amount)
class UPI_payment(payment):
    def process_payment(self):
        print('processing UPI payment of amount:',self.amount)
class net_banking_payment(payment):
    def process_payment(self):
        print('processing net banking payment of amount:',self.amount)

def make_payment(payment_method):
    payment_method.process_payment()

c=credit_card_payment(1000)
u=UPI_payment(200)
n=net_banking_payment(30000)
make_payment(c)
make_payment(u) 
make_payment(n)
```

```python
class BankAccount:
    def __init__(self,__account_number, __balance):
        self.__account_number = __account_number
        self.__balance = __balance
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f'Deposited {amount}. New balance: {self.__balance}')
    def withdraw(self, amount):
        if amount > 0 and amount <= self.__balance:
            self.__balance -= amount
            print(f'Withdrew {amount}. New balance: {self.__balance}')
        else:
            print('Insufficient funds.')
    def get_balance(self):
        return f'Current balance: {self.__balance}'
    def get_account_number(self):
        return f'Account numbers is: {self.__account_number}'

acc=BankAccount('123456789', 1000)
acc.deposit(500)
acc.withdraw(200)
acc.withdraw(1500)
print(acc.get_account_number())
print(acc.get_balance())
```

```python
from abc import ABC, abstractmethod
class employee(ABC):
    def __init__(self,name):
        self.name=name
        self.salary=0
    def display_details(self):
        print('employee name:',self.name)
        print('employee salary:',self.salary)
    @abstractmethod
    def calculate_salary(self):
        pass

class Full_time_employee(employee):
    def __init__(self,name,monthly_salary):
        super().__init__(name)
        self.monthly_salary=monthly_salary
    def calculate_salary(self):
        self.salary=self.monthly_salary
        print(self.salary)
class Part_time_employee(employee):
    def __init__(self,name,hours_worked, hourly_rate):
        super().__init__(name)
        self.hours_worked=hours_worked
        self.hourly_rate=hourly_rate
    def calculate_salary(self):
        self.salary=self.hours_worked*self.hourly_rate
        print(self.salary)

ff=Full_time_employee('shiva', 5000)
pp=Part_time_employee('karan', 20, 15)
ff.calculate_salary()
pp.calculate_salary()
```

```python
n=12345
def sum_of_digits(n):
    s=0
    while n > 0:
        r=n%10
        s+=r
        n//=10
    return s

while n > 9:
    n=sum_of_digits(n)
print(n)
```
