```python
class cars:
    def __init__(self,car_name,car_colour,car_price,car_milage,current_speed):
        self.car_name=car_name
        self.car_colour=car_colour
        self.car_price=car_price    
        self.car_milage=car_milage
        self.current_speed=current_speed
    def info(self):
        print('car name is :',self.car_name)
        print('car colour is :',self.car_colour)
        print('car price is :',self.car_price)
        print('car milage is :',self.car_milage)
        print('current speed is :',self.current_speed)
    def start(self):
        print(f'{self.car_name} is started.....')
    def accelerate(self,current_speed):
        if current_speed >= 100: 
            print(f'{self.car_name} is at maximum speed')
        else:
            self.current_speed += 10
            print(f'{self.car_name} is accelerating, current speed is {self.current_speed}')
    def brake(self,current_speed):
        if self.current_speed == 0:
            print(f'{self.car_name} is stoped')
        else:
            self.current_speed -= 10
            print(f'{self.car_name} is decelerating, current speed is {self.current_speed}') 
    def stop(self):
        print(f'{self.car_name} is stopped.....')

c2=cars('BMW','black',5000000,10,50)
c2.info()
c2.start()
c2.accelerate(c2.current_speed)
c2.accelerate(c2.current_speed)
c2.accelerate(c2.current_speed)
c2.accelerate(c2.current_speed)
c2.accelerate(c2.current_speed)
c2.accelerate(c2.current_speed)
c2.brake(c2.current_speed)
c2.brake(c2.current_speed)  
c2.brake(c2.current_speed)
c2.brake(c2.current_speed)
c2.brake(c2.current_speed)
c2.brake(c2.current_speed)
c2.stop()

```

```python
class Basic_calculator:
    def add(self,a,b):
        print(a+b)
    def subtract(self,a,b):
        print(a-b)  
    def multiply(self,a,b):
        print(a*b)
class Scientific_calculator(Basic_calculator):
    def power(self,a,b):
        print(a**b)
    def square_root(self,a):
        print(a**0.5)
class advanced_calculator(Scientific_calculator):
    def modulus(self,a,b):
        print(a%b)

B=advanced_calculator()
B.add(10,5)
B.subtract(10,5)
B.multiply(10,5)
B.power(10,5)
B.square_root(25)
B.modulus(10,3)
```

Types of inheritance

Single-level inheritance

```python
class parent:
    def sleep(self):
        print('sleeping.....')
    def eat(self):
        print('eating.....')
class child(parent):
    def play(self):
        print('playing.....')
c=child()
c.sleep()
c.eat()
c.play()
```

Multi-level inheritance

```python
class grandparent:
    def sleep(self):
        print('sleeping.....')
class parent(grandparent):
    def work(self):
        print('working.....')
    def eat(self):
        print('eating.....')
class child(parent):
    def play(self):
        print('playing.....')
c=child()
c.sleep()
c.work()
c.eat()
c.play()
```

Hierarchical inheritance

```python
class mother:
    def sleep(self):
        print('sleeping.....')
class son(mother):
    def play(self):
        print('playing.....')
class daughter(mother):
    def read(self):
        print('reading.....')
s=son()
s.sleep()
s.play()
d=daughter()
d.sleep()
d.read()
```

Mulptiple inheritance

```python
class Father:
    def work(self):
        print('working.....')   
class mother:
    def sleep(self):
        print('sleeping.....')
class son(Father, mother):
    def play(self):
        print('playing.....')

s=son()
s.work()
s.sleep()
s.play()
```

Hybride inheritance

```python
class A:
    def method_a(self):
        print('method a is called')
class B(A):
    def method_b(self):
        print('method b is called')
class C(A):
    def method_c(self):
        print('method c is called')
class D(B,C):
    def method_d(self):
        print('method d is called')
d=D()
d.method_a()
d.method_b()
d.method_c()    
d.method_d()
```

Duck typing

```python
class Robo:
    def Run(self):
        print('Robo is running.....')
    def Jump(self):
        print('Robo is jumping.....')
class FighterRobo(Robo):
    pass
class teacherRobo(Robo):
    pass
class driverRobo(Robo):
    pass
def main(Acction):
    Acction.Run()
    Acction.Jump()
Acction=FighterRobo()
main(Acction)
Acction=teacherRobo()
main(Acction)
Acction=driverRobo()    
main(Acction)
```

Overloading Default parameters

```python
class calculator:
    def add(self,a=0):
        print(a)
    def add(self,a=0,b=0):
        print(a+b)
    def add(self,a=0,b=0,c=0):
        print(a+b+c)
    def add(self,a=0,b=0,c=0,d=0):
        print(a+b+c+d)


c=calculator()
c.add(1)
c.add(1,2)
c.add(1,2,3)
c.add(1,2,3,4)
```

Overloading with * args


```python
class calculator:
    def add(self,*num):
        # print(sum(num))
        sum=0
        for i in num:
            sum+=i  
        print(sum)
c=calculator()
c.add(10)
c.add(10,20)
c.add(10,20,30)
c.add(10,20,30,40)
```

Overriding

```python
class Robo:
    def Run(self):
        print('Robo is running.....')
    def action(self):
        print('Robo is performing an action.....')
class FighterRobo(Robo):
    def action(self):
        print('FighterRobo is performing an action.....')
class teacherRobo(Robo):
    def action(self):
        print('teacherRobo is performing an action.....')
def main(A):
    A.Run()
    if isinstance(A, FighterRobo):
        A.action()
    elif isinstance(A, teacherRobo):
        A.action()
main(FighterRobo())
main(teacherRobo())
```

Abstract class

```python
from abc import ABC, abstractmethod
class calculator(ABC):
    @abstractmethod
    def addtion(self,a,b):
        pass
    def subraction(self,a,b):
        print(f'subraction of {a} and {b} is {a-b} ') 
class updated_calculator(calculator):
    def addtion(self,a,b):
        print(f'addtion of {a} and {b} is {a+b} ') 
    def divide(self,a,b):
        print(f'Division of {a} and {b} is {a/b} ') 
c=updated_calculator()
c.addtion(10,5)
c.subraction(10,5)
c.divide(10,5)
```

constructor Overloading

```python
class calculator:
    def __init__(self,*num):
        if len(num) == 0:
            print('Default constructor')
        elif len(num) == 1:
            self.a=num[0]
            print(self.a)
        elif len(num) == 2:
            self.a=num[0]
            self.b=num[1]
            print(self.a+self.b)
        elif len(num) == 3:
            self.a=num[0]
            self.b=num[1]
            self.c=num[2]
            print(self.a+self.b+self.c)
        else:
            print(sum(num))

calculator()
calculator(10)
calculator(10,20)
calculator(12,14,16)
calculator(10,20,30,40)
```

```python
class numbers:
    def __init__(self,*num):
        if len(num) == 0:
            print('Default constructor')
        else:
            print(sum(num))

numbers()
numbers(10)
numbers(10,20)
numbers(12,14,16)
numbers(10,20,30,40)
```
