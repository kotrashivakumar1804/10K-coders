TYPES OF EXCEPTIONS

Generic Exception

```python
a=5
b=0
try:
    print(a//b) 
except Exception as e:
    print('Exception handled')
```

Specific Exception

```python
a=6
b=0
try:
    print(a//b)
except ZeroDivisionError as e:
    print('cannot divide by zero')
```

Trying with different combinations

```python
a=5
b=10
try:
    c=a//b
except ZeroDivisionError as e:
    print('cannot divide by zero')
else:
    print(c)
finally:
    print('Thank you')
```

```python
a=6
b=0
try:
    c=a//b
    try:
        print("I am inner try block")
    except Exception as e:
        print("I am inner except block")
    else:
        print("I am inner else block")
    finally:
        print("I am inner finally block")
except ZeroDivisionError as e:
    print('cannot divide by zero')
else:
    print(c)
finally:
    print('Thank you')


    
```

```python
a=6
b=0
try:
    print("i am outer try block")
    try:
        c=a//b  
    except Exception as e:
        print("exception handled in inner except block")
    else:
        print(c)
    finally:
        print("I am inner finally block")
except ZeroDivisionError as e:
    print('cannot divide by zero')
else:
    print("I am outer else block")
finally:
    print('I am outer finally block')
```

DIFFERENT TYPES OF FUNCTIONS

Higher order function   

----> A function takes another function as a argement (or) Return anothet function is called Higher order function

```python
def vechile(name):
    print(f'I want to buy a {name} car')
def car(fun):
    fun('BMW')

car(vechile)
```

Callback Function

--------->A function Acting as argement to another function is called callback function

```python
def flower(name):
    print(f'I have a {name} flower')
def greet(fun):
    for i in range(1,4):
        fun('rose')
        
greet(flower)
```

Lambda Function

-----> anonymous single line function is called lambda function

normal function

```python
def position(a):
    if a % 2 == 0:
        print('It is even number')
    else:
        print('It is odd number')

position(5)
position(6)
```

lambda function

```python
L=lambda a:'Even number' if a%2 == 0 else 'Odd number'
print(L(5))
print(L(6))
```

CLOSURE FUNCTION  

-----> inner function remembers the value of x even after outer function execuation completed

```python
def outer():
    x=10
    def inner():
        print(x)
    return inner
o=outer()
o()
```

SUPER ( ) METHOD  

------>super() method is used to access features from the parent class into the child class, reuse them, and add new features in the child class.

```python
class vechile:
    def __init__(self,brand,model):
        self.brand=brand
        self.model=model
class car(vechile):
    def __init__(self,brand,model,speed,mileage):
        super().__init__(brand,model)
        self.speed=speed
        self.mileage=mileage
    
    def info(s):
        print(s.brand)
        print(s.model)
        print(s.speed)
        print(s.mileage)

c=car('BMW','X5',200,15)
c.info()
```
