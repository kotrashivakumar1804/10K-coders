OPERATORS

```python
# Arthimatic operations
#-----------------------------
arr=np.array([1,2,3])
arr+5
arr-5
arr*2
arr/3
arr//6
arr%6
arr**2

arr1=np.array([3,4,5])
np.add(arr,arr1)
np.subtract(arr,arr1)
np.multiply(arr,arr1)
np.divide(arr,arr1)
np.floor_divide(arr,arr1)
np.mod(arr,arr1)
np.power(arr,arr1)

#Comperaction Operations
#-----------------------------  
arr==arr1
arr!=arr1
arr>arr1
arr<arr1
arr>=arr1
arr<=arr1

np.less_equal(arr,arr1)
np.greater_equal(arr,arr1)
np.less(arr,arr1)
np.greater(arr,arr1)
np.equal(arr,arr1)
np.not_equal(arr,arr1)
```
